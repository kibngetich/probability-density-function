```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import seaborn as sns
import matplotlib.pyplot as plt
plt.style.use("ggplot")
```

```python
df=pd.read_csv("weight-height.csv")
```

```python
fig,(ax1,ax2)=plt.subplots(2,sharex=True,gridspec_kw={"height_ratios": (.15, .85)}, figsize = (10,8))
sns.histplot(df.Height,
            alpha=0.4,
            edgecolor="b",
            color="w",
            label="histogram",
            linewidth=2,
            stat="density",
            ax=ax2)
sns.kdeplot(df.Height,
             alpha=0.7,
             linewidth=3,
             label="Kernel Density plot",
             color="r",
             ax=ax2
             )
mean=df.Height.mean()
std=df.Height.std()
parametric_distribution=stats.norm(loc=mean,scale=std)
x=np.linspace(parametric_distribution.ppf(0.01),parametric_distribution.ppf(0.99),100)
ax2.plot(x,
         parametric_distribution.pdf(x),
         linestyle="-",
         label="parametric fit",
         color="g",
         alpha=0.5)
# ax2.set_title("Density estimations")
# sns.boxplot(df.Height,ax=ax1,color="r")
ax1.set_title('Box and Whiskers Plot')
ax2.set(ylim=(0, 0.11))
plt.ylim(0, 0.11)
plt.legend();
```
